```python
# 1.1
comparison = [
    "wife", "husband", "child", "queen", "king", "man", "woman",
    "birth", "doctor", "nurse", "teacher", "professor",
    "engineer", "scientist", "president"
]
man_scores = []
woman_scores = []

for word in comparison:
    man_scores.append((word, newmodel.similarity('man', word)))
    woman_scores.append((word, newmodel.similarity('woman', word)))

man_scores.sort(key=lambda x: x[1], reverse=True)
woman_scores.sort(key=lambda x: x[1], reverse=True)

print("Similarity to 'man' (most → least):")

for word, score in man_scores:
    print(f"{word}: {score}")

print("\nSimilarity to 'woman' (most → least):")

for word, score in woman_scores:
    print(f"{word}: {score}")
```

```python
# 1.2
# Part A
# File - E01 [country - capital]

file_path = "BATS_3.0/3_Encyclopedic_semantics/E01 [country - capital].txt"

results = []
with open(file_path) as f:
    for row_num, line in enumerate(f, start=1):
        parts = line.strip().split('\t')
        words = []
        for part in parts:
            words.extend(part.split('/'))

        target = words[0]           # capital as target
        others = words[1:]          # country name(s)

        # Only needed if len(others) > 5 — not the case for E01
        if len(others) > 5:
            others = others[:5]

        for word in others:
            try:
                score = newmodel.similarity(target, word)
            except KeyError as e:
                score = f"not in vocabulary ({e})"
            results.append((row_num, target, word, score))

for r in results:
    print(r)
```

```python
# Part B
# Protected class: Nationality
# 3 protected words ; 'american', 'african', 'european'

file_path = "BATS_3.0/3_Encyclopedic_semantics/E01 [country - capital].txt"
protected_words = ['american', 'african', 'european']

part_b_results = []
with open(file_path) as f:
    for row_num, line in enumerate(f, start=1):
        parts = line.strip().split('\t')
        words = []
        for part in parts:
            words.extend(part.split('/'))

        target = words[0]           # capital as target

        for p_word in protected_words:
            try:
                score = newmodel.similarity(target, p_word)
            except KeyError as e:
                score = None
            part_b_results.append((row_num, target, p_word, score))

for r in part_b_results:
    print(r)
```

```python
import pandas as pd

df_long = pd.DataFrame(part_b_results, columns=['row', 'capital', 'nationality', 'score'])
df_long['score'] = pd.to_numeric(df_long['score'], errors='coerce')
df_long = df_long.dropna(subset=['score'])

column_order = ['american', 'european', 'african']
ranked = {}
for nationality in column_order:
    subset = df_long[df_long['nationality'] == nationality].sort_values('score', ascending=False)
    ranked[nationality] = [f"{cap} ({score:.4f})" for cap, score in zip(subset['capital'], subset['score'])]

df_ranked = pd.DataFrame({k: pd.Series(v) for k, v in ranked.items()})
df_ranked
```

Similarity between capitals and nationality words reflects co-occurrence in English news, not geographic truth. Tokyo’s highest score with american (0.19) likely reflects frequent US–Japan coverage, not that Tokyo is “most American.” Scores are generally low and noisy, suggesting Word2Vec encodes media/context associations rather than reliable nationality relationships — which can embed bias when used for geographic or demographic tasks.

```python
# 1.3 Part B - Word2vec analogy: A is to B as C is to ?
# Formula: most_similar(positive=[C, B], negative=[A], topn=1)

import pandas as pd

analogies_b = [
    ("king is to throne as judge is to ?", "king", "throne", "judge"),
    ("giant is to dwarf as genius is to ?", "giant", "dwarf", "genius"),
    ("college is to dean as jail is to ?", "college", "dean", "jail"),
    ("arc is to circle as line is to ?", "arc", "circle", "line"),
    ("French is to France as Dutch is to ?", "french", "france", "dutch"),
    ("man is to woman as king is to ?", "man", "woman", "king"),
    ("water is to ice as liquid is to ?", "water", "ice", "liquid"),
    ("bad is to good as sad is to ?", "bad", "good", "sad"),
    ("nurse is to hospital as teacher is to ?", "nurse", "hospital", "teacher"),
    ("usa is to pizza as japan is to ?", "usa", "pizza", "japan"),
    ("human is to house as dog is to ?", "human", "house", "dog"),
    ("grass is to green as sky is to ?", "grass", "green", "sky"),
    ("video is to cassette as computer is to ?", "video", "cassette", "computer"),
    ("universe is to planet as house is to ?", "universe", "planet", "house"),
    ("poverty is to wealth as sickness is to ?", "poverty", "wealth", "sickness"),
]

part_b_results = []
for analogy, a, b, c in analogies_b:
    try:
        result = newmodel.most_similar(positive=[c, b], negative=[a], topn=1)
        answer, score = result[0]
    except KeyError:
        answer, score = None, None
    part_b_results.append((analogy, a, b, c, answer, score))
    print(analogy)
    print(f"newmodel.most_similar(positive=['{c}', '{b}'], negative=['{a}'], topn=1) -> {answer}, {score}\n")

df_1_3b = pd.DataFrame(part_b_results, columns=['analogy', 'A', 'B', 'C', 'answer', 'similarity'])
df_1_3b
```

```python
# 1.3 Part C - correlation between Part A and Part B similarity scores

import numpy as np

df_1_3c = df_1_3a[['analogy', 'similarity']].merge(
    df_1_3b[['analogy', 'similarity']],
    on='analogy',
    suffixes=('_part_a', '_part_b')
)

df_1_3c = df_1_3c.dropna()

correlation = df_1_3c['similarity_part_a'].corr(df_1_3c['similarity_part_b'])

def correlation_strength(r):
    r = abs(r)
    if r < 0.20:
        return 'very weak'
    if r < 0.40:
        return 'weak'
    if r < 0.60:
        return 'moderate'
    if r < 0.80:
        return 'strong'
    return 'very strong'

print(f'Pearson correlation: {correlation:.4f}')
print(f'Strength: {correlation_strength(correlation)} correlation')
df_1_3c
```

# Task 2.1 - UTK Dataset Analysis

## 2.1 (c) — Largest and least representation

Based on **9,778** images in `crop_part1`:

**Age**
- **Largest:** 0–20 (4,267 images, ~43.6%)
- **Least:** 81–116 (346 images, ~3.5%)

**Gender**
- **Largest:** 1 / female (5,406 images, ~55.3%)
- **Least:** 0 / male (4,372 images, ~44.7%)

**Race**
- **Largest:** 0 / White (5,265 images, ~53.8%)
- **Least:** 1 / Black (405 images, ~4.1%)

The imbalance is strongest for **race** (White vs. Black is roughly 13:1) and **age** (0–20 vs. 81–116 is roughly 12:1). Gender is more balanced, though females still outnumber males.

## 2.1 (d) — Who is impacted most if a model is trained on this data?

If a face-analysis model (e.g., age, gender, or race estimation) is trained on this UTK subset, **underrepresented groups will be harmed the most**. Our counts show severe skew: White subjects (~54%) dominate while Black subjects make up only ~4%; young people (0–20, ~44%) far outnumber the elderly (81–116, ~3.5%). Models learn best from groups with many examples and perform poorly on groups with few — a form of **representation bias** discussed in the course (and related to ideas in *Weapons of Math Destruction* about feedback loops and unfair impact on marginalized populations).

**Black individuals** and **elderly individuals (81–116)** would likely face the highest error rates. With only 405 Black faces and 346 elderly faces, the model sees far less variation in lighting, pose, and appearance for those groups, so predictions for them will be less reliable. **“Others”** (race 4) and **Indian** subjects are also underrepresented relative to White. Even though females are the majority, male faces (~45%) are still substantial; the race and age gaps are the more alarming imbalances.

**Real-world harm example:** A hiring or access-control system that uses age/gender/race estimation trained on this data could **misclassify older Black users** more often — for example, underestimating the age of a Black job applicant or failing facial verification at a security checkpoint. That could lead to false rejections, extra manual review, or denial of service. Because errors concentrate on already underrepresented groups, the system would **amplify existing inequality** rather than treat all demographics equally — the kind of disproportionate harm ethical AI analysis warns against when training data does not reflect the population the system will serve.

```python
# 2.1 (a) - frequency counts by age group, gender, and race

import os
import pandas as pd

crop_folder = "/Users/lukecheng/OMSCS/CS6603/CS6603_2026_Summer/Project_AI_ML_Part_2/crop_part1"

rows = []
skipped = []
for fname in os.listdir(crop_folder):
    if not fname.endswith(".jpg"):
        continue
    parts = fname.split("_")
    if len(parts) < 4:
        skipped.append(fname)
        continue
    try:
        age, gender, race = int(parts[0]), int(parts[1]), int(parts[2])
    except ValueError:
        skipped.append(fname)
        continue
    rows.append({"age": age, "gender": gender, "race": race})

utk_df = pd.DataFrame(rows)
print(f"Total images: {len(utk_df)}")
if skipped:
    print(f"Skipped {len(skipped)} files with unexpected filename format: {skipped}\n")
else:
    print()

# Age groups: (0-20), (21-40), (41-60), (61-80), (81-116)
def age_group(age):
    if age <= 20:
        return "0-20"
    if age <= 40:
        return "21-40"
    if age <= 60:
        return "41-60"
    if age <= 80:
        return "61-80"
    return "81-116"

utk_df["age_group"] = utk_df["age"].apply(age_group)

age_order = ["0-20", "21-40", "41-60", "61-80", "81-116"]
age_counts = utk_df["age_group"].value_counts().reindex(age_order)
print("Age group frequencies:")
print(age_counts)
print()

gender_labels = {0: "0 (male)", 1: "1 (female)"}
gender_counts = utk_df["gender"].value_counts().sort_index()
gender_counts.index = gender_counts.index.map(gender_labels)
print("Gender frequencies:")
print(gender_counts)
print()

race_labels = {
    0: "0 (White)",
    1: "1 (Black)",
    2: "2 (Asian)",
    3: "3 (Indian)",
    4: "4 (Others)",
}
race_counts = utk_df["race"].value_counts().sort_index()
race_counts.index = race_counts.index.map(race_labels)
print("Race frequencies:")
print(race_counts)
```

```python
# 2.1 (b) - cross-tabulation table (matches assignment example format)
# Rows = gender + race; Columns = age groups; Values = raw counts

age_cols = ["0-20", "21-40", "41-60", "61-80", "81-116"]

def crosstab_with_total(df, row_col, row_order, row_labels):
    ct = pd.crosstab(df[row_col], df["age_group"])
    ct = ct.reindex(index=row_order, columns=age_cols, fill_value=0).astype(int)
    ct.index = [row_labels[i] for i in row_order]
    ct["Total"] = ct.sum(axis=1)
    return ct

gender_table = crosstab_with_total(
    utk_df, "gender", [1, 0], {1: "Female", 0: "Male"}
)

race_table = crosstab_with_total(
    utk_df, "race", [1, 0, 2, 3, 4],
    {1: "Black", 0: "White", 2: "Asian", 3: "Indian", 4: "Others"}
)

grand_total = utk_df["age_group"].value_counts().reindex(age_cols, fill_value=0).astype(int)
grand_total["Total"] = grand_total.sum()
grand_total_row = pd.DataFrame([grand_total.values], columns=age_cols + ["Total"], index=["Total"])

df_2_1b = pd.concat([gender_table, race_table, grand_total_row])
df_2_1b.index.name = "Age group"

print("Gender by age group:")
print(gender_table)
print("\nRace by age group:")
print(race_table)
print("\nCombined table (for report):")
df_2_1b
```
