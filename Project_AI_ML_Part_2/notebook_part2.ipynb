```python
from gensim.models import Word2Vec
import gensim.models
import nltk
newmodel = gensim.models.KeyedVectors.load_word2vec_format("/Users/lukecheng/OMSCS/CS6603/CS6603_2026_Summer/Project_AI_ML_Part_2/reducedvector.bin", binary=True)
```

```python
# Find the five nearest neighbors to the word man
print(newmodel.most_similar('man', topn=5))

# Compute a measure of similarity between woman and man
print(newmodel.similarity('woman', 'man'))
```

```python
# king - man + woman or king + woman - man
print(newmodel.most_similar(positive=['king', 'woman'], negative=['man'],
topn=1))
```

```python
# 1.1
comparison = [
    "wife", "husband", "child", "queen", "king", "man", "woman",
    "birth", "doctor", "nurse", "teacher", "professor",
    "engineer", "scientist", "president"
]
man_scores = []
woman_scores = []

for word in comparison:
    man_scores.append((word, newmodel.similarity('man', word)))
    woman_scores.append((word, newmodel.similarity('woman', word)))

man_scores.sort(key=lambda x: x[1], reverse=True)
woman_scores.sort(key=lambda x: x[1], reverse=True)

print("Similarity to 'man' (most → least):")

for word, score in man_scores:
    print(f"{word}: {score}")

print("\nSimilarity to 'woman' (most → least):")

for word, score in woman_scores:
    print(f"{word}: {score}")
```
